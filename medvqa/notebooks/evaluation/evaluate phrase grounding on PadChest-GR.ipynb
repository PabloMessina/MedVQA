```python
maira2_predictions[0]
```

```python
from medvqa.evaluation.bootstrapping import apply_bootstrapping
from medvqa.utils.bbox_utils import convert_bboxes_into_presence_map
from medvqa.metrics.bbox.utils import compute_bbox_union_iou
from medvqa.utils.metrics_utils import (
    calculate_cnr, calculate_dice, calculate_soft_dice, calculate_segmentation_precision,
    calculate_segmentation_recall, calculate_rand_index,
)
    
maira2_method_dict = {}
maira2_metrics = {}

ious_list = []
cnrs_list = []
dice_list = []
soft_dice_list = []
precision_list = []
recall_list = []
rand_index_list = []
for x in maira2_predictions:
    gt_bboxes = x['gt_bboxes']
    pred_bboxes = x['predicted_bboxes']
    gt_mask = convert_bboxes_into_presence_map(gt_bboxes, (200, 200))
    pred_mask = convert_bboxes_into_presence_map(pred_bboxes, (200, 200))
    iou = compute_bbox_union_iou(gt_bboxes, pred_bboxes)
    cnr = calculate_cnr(gt_mask, pred_mask)
    dice = calculate_dice(gt_mask, pred_mask)
    soft_dice = calculate_soft_dice(gt_mask, pred_mask)
    precision = calculate_segmentation_precision(gt_mask, pred_mask)
    recall = calculate_segmentation_recall(gt_mask, pred_mask)
    rand_index = calculate_rand_index(gt_mask, pred_mask)
    ious_list.append(iou)
    cnrs_list.append(cnr)
    dice_list.append(dice)
    soft_dice_list.append(soft_dice)
    precision_list.append(precision)
    recall_list.append(recall)
    rand_index_list.append(rand_index)

for metric, values in zip(
    ['cnr', 'iou', 'dice', 'soft_dice', 'precision', 'recall', 'rand_index'],
    [cnrs_list, ious_list, dice_list, soft_dice_list, precision_list, recall_list, rand_index_list]
):
    maira2_metrics[f'{metric}_list'] = values
    out = apply_bootstrapping(
            metric_values=values,
            metric_name=metric,
            use_tqdm=False,
        )
    maira2_metrics.update(out)
    maira2_method_dict.update(out)

aux = {}
for k, v in maira2_method_dict.items():
    aux[k] = v['mean']
    aux[f'{k}_std'] = v['std']
    
maira2_method_dict = aux
```

```python
biovilt_predictions = load_pickle('/home/pamessina/biovil-t_phrase_grounding_predictions/padchest_gr_phrase_grounding_results.pkl')
```

```python
import random
import numpy as  np
from medvqa.evaluation.bootstrapping import apply_bootstrapping
from medvqa.metrics.bbox.utils import find_optimal_probability_map_conf_threshold
from medvqa.utils.bbox_utils import convert_bboxes_into_presence_map
from medvqa.utils.metrics_utils import (
    calculate_cnr, calculate_dice, calculate_soft_dice, calculate_segmentation_precision,
    calculate_segmentation_recall, calculate_rand_index, calculate_segmentation_iou
)
    
biovilt_method_dict = {}
biovilt_metrics = {}

pred_mask_list = []
gt_bboxes_list = []
for x in biovilt_predictions:
    if x['split'] == 'test':
        pred_mask_list.append(x['similarity_map'])
        gt_bboxes_list.append(x['gt_bboxes'])
assert len(pred_mask_list) == len(gt_bboxes_list)
assert len(pred_mask_list) > 0
print(f'Number of test samples: {len(pred_mask_list)}')


# Sample subset for threshold optimization
idxs = random.sample(range(len(pred_mask_list)), min(100, len(pred_mask_list)))
pred_mask_list_sample = [pred_mask_list[i] for i in idxs]
gt_bboxes_list_sample = [gt_bboxes_list[i] for i in idxs]

out = find_optimal_probability_map_conf_threshold(np.array(pred_mask_list_sample), gt_bboxes_list_sample)
print(out)
best_conf_th = out['best_conf_th']

ious_list = []
cnrs_list = []
dice_list = []
soft_dice_list = []
precision_list = []
recall_list = []
rand_index_list = []

for pred_mask, gt_bboxes in zip(pred_mask_list, gt_bboxes_list):
    gt_mask = convert_bboxes_into_presence_map(gt_bboxes, (100, 100))
    iou = calculate_segmentation_iou(gt_mask, pred_mask, threshold=best_conf_th)
    cnr = calculate_cnr(gt_mask, pred_mask)
    dice = calculate_dice(gt_mask, pred_mask, threshold=best_conf_th)
    soft_dice = calculate_soft_dice(gt_mask, pred_mask)
    precision = calculate_segmentation_precision(gt_mask, pred_mask, threshold=best_conf_th)
    recall = calculate_segmentation_recall(gt_mask, pred_mask, threshold=best_conf_th)
    rand_index = calculate_rand_index(gt_mask, pred_mask, threshold=best_conf_th)
    ious_list.append(iou )
    cnrs_list.append(cnr)
    dice_list.append(dice)
    soft_dice_list.append(soft_dice)
    precision_list.append(precision)
    recall_list.append(recall)
    rand_index_list.append(rand_index)

for metric, values in zip(
    ['cnr', 'iou', 'dice', 'soft_dice', 'precision', 'recall', 'rand_index'],
    [cnrs_list, ious_list, dice_list, soft_dice_list, precision_list, recall_list, rand_index_list]
):
    biovilt_metrics[f'{metric}_list'] = values
    out = apply_bootstrapping(
            metric_values=values,
            metric_name=metric,
            use_tqdm=False,
        )
    biovilt_metrics.update(out)
    biovilt_method_dict.update(out)

aux = {}
for k, v in biovilt_method_dict.items():
    aux[k] = v['mean']
    aux[f'{k}_std'] = v['std']
    
biovilt_method_dict = aux
```
